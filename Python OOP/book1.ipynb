```python
class Book():

    def __init__(self,title,author,year,genre):
        self.title = title
        self.author = author
        self.year = year
        self.genre = genre

    def __str__(self):
        return (
            'Title: ' + self.title + ', '+
                'Author: ' + self.author +', '+
                'Year: ' + str(self.year) +', '+
                'Genre: ' + ','+str(self.genre)
                )
    @staticmethod
    def load_from_file_line(line):
        parts = line.split(',',3)

        for i in range(len(parts)):
           parts[i] = parts[i].strip()

        parts[2] = int(parts[2])
        parts[3] = parts[3].strip('[]')
        parts[3] = parts[3].split(',')
        parts[3] = parts[3].strip()



line = Book('The Hobbit', 'J.R.R Tolkien', 1937, ['Fantasy','Adventure'])
print(line)
```
