```python

def longest_substring_with_k(s, num):
    longest = ''
    current = ''
    char_count = {}

    for i in s:
        current += i

        if i in char_count:
            char_count[i] += 1
        else:
            char_count[i] = 1

        while len(char_count) > num:
            first = current[0]
            char_count[first] -= 1
            current = current[1:]
            
            if char_count[first] == 0:
                del char_count[first]

        if len(char_count) == num:
            if len(current) > len(longest):
                longest = current

    return longest



print(longest_substring_with_k("aabbbcc", 2))
print(longest_substring_with_k("banana", 2))
print(longest_substring_with_k("abaccc", 2))
print(longest_substring_with_k("mississippi", 2))
print(longest_substring_with_k("", 2))
```
