```python
import random

class Card:
   def __init__(self,rank,suit):
      self.suit = suit
      self.rank = rank
   def __str__(self):
      return f'{self.rank} of {self.suit}'

class Deck:

   def __init__(self):
      self.cards = []
      
   def create_standard_deck(self, include_joker = False):

      
      # ensures deck is empty before adding 52 to it
      self.cards = []
      

      suits = ['Hearts', 'Diamonds', 'Clubs', 'Spades']
      ranks = ['Ace', '2', '3', '4', '5', '6', '7',
               '8', '9', '10', 'Jack', 'Queen', 'King']

      # creates deck 
      for suit in suits:
         for rank in ranks:
            self.cards.append(Card(rank,suit))

      # checks if joker is needed and adds it
      if include_joker:
            self.cards.append(Card("Joker", "None"))

   def shuffle(self):
      random.shuffle(self.cards)
   
   def deal(self,number):
      dealt = self.cards[0:number]
      del self.cards[0:number]
      return dealt
   
class Player:
   def __init__(self,name):
      self.name = name 
      self.hand = []

   def add_card(self,card):
      self.hand.append(card) 

   def remove_card(self,card):
      self.hand.remove(card)

   def draw_card(self, deck):
      deck = deck.deal(1)[0]   
      self.add_card(deck)
   
   def draw_cards(self, deck, number):
      for i in range(number):
         self.draw_card(deck) 





class Game:
   
   def __init__(self, players):
      self.players = players
      self.current_player_index = 0
      self.direction = 1
      self.discard_pile = []

   def get_current_player(self):
      return self.players[self.current_player_index]
   
   def next_turn(self):
      self.current_player_index = (self.current_player_index + self.direction) % len(self.players)
      
   def jack_reverse(self):
      self.direction = self.direction * -1

   def skip_player(self):
      self.current_player_index = (self.current_player_index + (self.direction*2)) % len(self.players)

   
   # Deals certain number of cards to each player
   def deal_initial_hands(self,deck,number):
      for i in self.players:
         i.draw_cards(deck,number)

   def start_discard_pile(self,deck):
      self.discard_pile.append(deck.deal(1)[0])

   def get_top_discard(self):
      return self.discard_pile[-1]

   def can_play_card(self,card):

      top_card = self.get_top_discard()

      if top_card.rank == card.rank or top_card.suit == card.suit:
         return True
      else:
         return False
   def play_card(self,card):
      if self.can_play_card(card) == True:
         current_player = self.get_current_player()
         current_player.remove_card(card)
         self.discard_pile.append(card)
      else:
         print('You can not play that card.')
         



player1 = Player("Alice")
player2 = Player("Bob")
player3 = Player("Charlie")

players = [player1, player2, player3]

game = Game(players)

deck = Deck()
deck.create_standard_deck()
deck.shuffle()

game.deal_initial_hands(deck, 7)
game.start_discard_pile(deck)

print("Starting discard card:", game.discard_pile[-1])
print("Cards in discard pile:", len(game.discard_pile))
print("Cards remaining in deck:", len(deck.cards))

top_card = game.get_top_discard()
print("Top card:", top_card)

# Test matching rank
test_card1 = Card(top_card.rank, "Hearts")
print(game.can_play_card(test_card1))

# Test matching suit
test_card2 = Card("King", top_card.suit)
print(game.can_play_card(test_card2))

# Test neither matching
test_card3 = Card("3", "Clubs")
print(game.can_play_card(test_card3))


print(game.get_top_discard())
```
