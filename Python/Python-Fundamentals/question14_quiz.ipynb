```python
class Person:
    def __init__(self, dob_day, dob_month, dob_year):
        self._dob_day = dob_day
        self._dob_month = dob_month
        self._dob_year = dob_year

    def __lt__(self,other):
        if self._dob_year < other._dob_year:
            return False
        elif self._dob_year > other._dob_year:
            return True   
        elif self._dob_year == other._dob_year:
            if self._dob_month < other._dob_month:
                return False
            elif self._dob_month > other._dob_month:
                return True
            elif self._dob_month == other._dob_month:
                if self._dob_day < other._dob_day:
                    return False
                elif self._dob_day > other._dob_day:
                    return True
                elif self._dob_day == other._dob_day:
                    return False

        

    



# mary = Person(30, 3, 1967)
# george = Person(30, 3, 1970)
# print(mary < george)  # Expected: False

# mary = Person(20, 2, 1967)
# george = Person(30, 3, 1967)
# print(mary < george)  # Expected: True

# Test 1: Different years
person1 = Person(10, 5, 2000)
person2 = Person(10, 5, 2005)
print(person1 < person2)  # Expected: False

# Test 2: Different years, reverse order
person1 = Person(10, 5, 2005)
person2 = Person(10, 5, 2000)
print(person1 < person2)  # Expected: True

# Test 3: Same year, different months
person1 = Person(15, 8, 2000)
person2 = Person(15, 6, 2000)
print(person1 < person2)  # Expected: True

# Test 4: Same year and month, different days
person1 = Person(20, 6, 2000)
person2 = Person(10, 6, 2000)
print(person1 < person2)  # Expected: True

# Test 5: Exactly the same birthday
person1 = Person(10, 5, 2000)
person2 = Person(10, 5, 2000)
print(person1 < person2)  # Expected: False
```
