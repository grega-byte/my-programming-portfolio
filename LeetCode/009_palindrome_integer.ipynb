```python
class Solution:
    def isPalindrome(self, x: int) -> bool:

        x = str(x)

        reverse = x[::-1]

        for i in range(len(x)):
            if x[i] != reverse[i]:
                return False 

        
        return True

solution = Solution()

print(solution.isPalindrome(121))       # True
print(solution.isPalindrome(-121))      # False
print(solution.isPalindrome(10))        # False
print(solution.isPalindrome(1221))      # True
print(solution.isPalindrome(12321))     # True
print(solution.isPalindrome(123))       # False
print(solution.isPalindrome(0))         # True
print(solution.isPalindrome(1001))      # True
print(solution.isPalindrome(100))       # False
print(solution.isPalindrome(1234321))   # True
```
