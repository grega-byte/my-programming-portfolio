```python
def containsDuplicate(nums):

    counter = {}

    for value in nums:
        if value in counter:
            return True
        else:
            counter[value] = 1
    return False


print(containsDuplicate([1, 2, 3, 1]))       # True
print(containsDuplicate([1, 2, 3, 4]))       # False
print(containsDuplicate([1, 1, 1, 3, 3]))    # True
print(containsDuplicate([5]))                # False
print(containsDuplicate([]))                 # False
print(containsDuplicate([1, 2, 3, 4, 5, 5])) # True
print(containsDuplicate([0, 0]))             # True
print(containsDuplicate([-1, -2, -3, -1]))  # True
```

```python
#an alternate version with sets

def containsDuplicate(nums):
    return len(set(nums)) != len(nums)

print(containsDuplicate([1, 2, 3, 1]))       # True
print(containsDuplicate([1, 2, 3, 4]))       # False
print(containsDuplicate([1, 1, 1, 3, 3]))    # True
print(containsDuplicate([5]))                # False
print(containsDuplicate([]))                 # False
print(containsDuplicate([1, 2, 3, 4, 5, 5])) # True
print(containsDuplicate([0, 0]))             # True
print(containsDuplicate([-1, -2, -3, -1]))  # True
```
