```python
def maxProfit(prices):

    if not isinstance(prices, list):
        raise TypeError('Prices must be a list.')
    elif len(prices) == 0:
        raise ValueError('Please enter numbers in your list.')

    
    lowest = prices[0]
    max_profit = 0

    for i in prices:

        if i < lowest:
            lowest = i
        profit = i - lowest
            
        if profit > max_profit:
            max_profit = profit
            
       
    return max_profit

    


print(maxProfit([7, 1, 5, 3, 6, 4]))  # 5
print(maxProfit([7, 6, 4, 3, 1]))    # 0
print(maxProfit([2, 4, 1]))          # 2
print(maxProfit([1, 2]))             # 1
print(maxProfit([2, 1]))             # 0
print(maxProfit([3, 3, 3, 3]))       # 0
print(maxProfit([1, 5, 2, 8]))       # 7
print(maxProfit([5, 1, 5, 1, 10]))   # 9



```
