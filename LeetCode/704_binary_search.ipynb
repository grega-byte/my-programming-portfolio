```python
def binary(nums,target):

    left = 0
    right = (len(nums)-1)
    

    

    while left<=right:
        calculate_mid = (left+right)//2

        if nums[calculate_mid] == target:
            return calculate_mid
        elif nums[calculate_mid] < target:
            left = calculate_mid+1
        elif nums[calculate_mid]>target:
            right = calculate_mid-1
        

    return -1




print(binary([-1, 0, 3, 5, 9, 12], 9))  # 4
print(binary([-1, 0, 3, 5, 9, 12], 2))  # -1
print(binary([5], 5))                    # 0
print(binary([1, 2, 3, 4, 5], 3))        # 2
print(binary([1, 2, 3, 4, 5], 5))        # 4
```
