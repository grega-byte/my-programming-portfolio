```python
def missingNumber(nums):

    expected_sum = sum(range(len(nums) + 1))
    
    total = sum(nums)
  
    
    return expected_sum - total
    
    

    




    

print(missingNumber([3, 0, 1]))                    # 2
print(missingNumber([0, 1]))                       # 2
print(missingNumber([9, 6, 4, 2, 3, 5, 7, 0, 1])) # 8
print(missingNumber([0]))                          # 1
print(missingNumber([1]))                          # 0
print(missingNumber([0, 2]))                       # 1
print(missingNumber([4, 2, 1, 0]))                # 3
print(missingNumber([2, 3, 0]))                    # 1
```
