```python
def isPalindrome(s):


    cleaned = []

    for i in s:
        if i.isalnum():
            cleaned += i.lower()

    cleaned = ''.join(cleaned)
    reverse = cleaned[::-1]
    

    for i in range(len(cleaned)):
        if cleaned[i] != cleaned[::-1][i]:
            return False
        
    return True
        



s = 'A man, a plan, a canal: Panama'
isPalindrome(s)
    
```
