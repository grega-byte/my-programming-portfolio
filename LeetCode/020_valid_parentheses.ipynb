```python
def brackets(s):

    stack = []

    pairs = {
    ')': '(',
    ']': '[',
    '}': '{'
    }

    for i in s:
        if i in '({[':
            stack.append(i)
        else:
            if len(stack) == 0:
                return False
            elif pairs[i] == stack[-1]:
                stack.pop()
            else:
                return False

    return len(stack) == 0
                       

print(brackets("()"))       # True
print(brackets("()[]{}"))   # True
print(brackets("(]"))       # False
print(brackets("([])"))     # True
print(brackets("([)]"))     # False
print(brackets("{[]}"))     # True
print(brackets("((("))      # False
print(brackets(""))         # True


```
