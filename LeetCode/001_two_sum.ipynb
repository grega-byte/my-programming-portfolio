```python
def two_sum(nums,target):  
    

    new_num = {}

    for index, key in enumerate(nums):
        needed = target - key
        
        if needed in new_num:
            return [new_num[needed],index]
        else:
            new_num[key]=index





print(two_sum([2, 7, 11, 15], 9))  # [1, 0] or [0, 1]
print(two_sum([3, 2, 4], 6))       # [2, 1] or [1, 2]
print(two_sum([3, 3], 6))          # [1, 0]




```
