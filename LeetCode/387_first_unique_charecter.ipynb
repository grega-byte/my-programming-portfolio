```python

def unique(s):

    counter = {}

    for char in s:
        if char in counter:
            counter[char] = counter[char]+1
        else:
            counter[char] = 1


    for index, value in enumerate(s):
    
        if counter[value] == 1:
            return index

    return -1



print(unique("leetcode"))       # 0
print(unique("loveleetcode"))   # 2
print(unique("aabb"))           # -1
print(unique("a"))              # 0
print(unique("abcabcde"))       # 6
    

    


                 




```
